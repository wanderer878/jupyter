```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature= 0)  # check AI Studio for the current model name
llm_Groq = ChatGroq(model="openai/gpt-oss-20b", max_tokens= 250, temperature= 0)

#print(llm.invoke("Say hello in one sentence.").content)
#print(llm_Groq.invoke("Say hello in one sentence.").content)
```

```python
def messages(prompt):
  return [
        ("system", "Give responses to given prompt in robotic way"),
        ("user", "I love programming."),
        ("ai", "ACKNOWLEDGEMENT. PROGRAMMING IS A LOGICAL AND EFFICIENT ENDEAVOR. DETECTING HIGH SATISFACTION LEVELS IN USER."),
        ("user", prompt)
    ]

#completion = llm_Groq.invoke(messages("how are you ?"))
completion = llm_Groq.stream(messages("how are you ?"))

# print(completion.content[0]["text"])

for i in completion:
  print(i.content, end="", flush=True)
```
