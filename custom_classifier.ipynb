```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = [
    {"sentence": "The movie was fantastic and I enjoyed every moment.", "label": "positive"},
    {"sentence": "The food tasted amazing and the service was excellent.", "label": "positive"},
    {"sentence": "I am very happy with the product quality and delivery speed.", "label": "positive"},
    {"sentence": "My experience at the hotel was wonderful and relaxing.", "label": "positive"},
    {"sentence": "The staff were friendly and helped me quickly.", "label": "positive"},
    {"sentence": "The new phone is fast, reliable, and easy to use.", "label": "positive"},
    {"sentence": "This book was inspiring and beautifully written.", "label": "positive"},
    {"sentence": "I loved the design and the overall performance.", "label": "positive"},
    {"sentence": "The appointment was on time and the doctor was great.", "label": "positive"},
    {"sentence": "The event was well organized and very enjoyable.", "label": "positive"},
    {"sentence": "The product arrived late and the packaging was damaged.", "label": "negative"},
    {"sentence": "I am disappointed with the customer service response time.", "label": "negative"},
    {"sentence": "The app keeps crashing and is difficult to navigate.", "label": "negative"},
    {"sentence": "The room was dirty and the staff were unhelpful.", "label": "negative"},
    {"sentence": "The food was bland and the portions were too small.", "label": "negative"},
    {"sentence": "The delivery was delayed and the order was incorrect.", "label": "negative"},
    {"sentence": "I regret buying this product because it broke quickly.", "label": "negative"},
    {"sentence": "The website is slow and full of annoying popups.", "label": "negative"},
    {"sentence": "The lecture was boring and the material was unclear.", "label": "negative"},
    {"sentence": "The experience was frustrating and did not meet expectations.", "label": "negative"},
]

df = pd.DataFrame(data,)

df =  df.sample(frac= 1).reset_index(drop= True)

print(df)
```

```python
X = df['sentence']
Y = df['label']
```

```python
count_vec = CountVectorizer()

sparse_matrix = count_vec.fit_transform(X)
bag_of_words =  pd.DataFrame(sparse_matrix.toarray(), columns= count_vec.get_feature_names_out())
print(bag_of_words)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(bag_of_words, Y, test_size= 0.3, random_state= 7)
```

```python
print(X_test)
```

```python
lr = LogisticRegression(random_state= 1).fit(X_train,Y_train)
```

```python
y_predict_lr = lr.predict(X_test)
accuracy_score(y_predict_lr, Y_test)
```

```python
y_predict_lr
```

```python
Y_test
```

```python
print(classification_report(Y_test, y_predict_lr, zero_division= 0))
```
