## Setting up the API

**Q1.** Import the required library and set up your API key so you can start making requests.

```python
from openai import OpenAI
from langchain_groq import ChatGroq
import httpx2
```

```python
import os
from pathlib import Path
from dotenv import load_dotenv

env_file = Path.cwd() / "secrets.env"
load_dotenv(env_file)

groq_key = os.getenv("GROQ_API_KEY")
openai_key = os.getenv("OPENAI_API_KEY")

if not groq_key:
    raise RuntimeError("GROQ_API_KEY is missing. Add it to secrets.env, then restart and run this cell.")
```

```python
http_client = httpx2.Client(headers={"Accept-Encoding": "gzip, deflate"})
client = OpenAI(
    base_url='https://api.groq.com/openai/v1/',
    http_client= http_client,
    api_key=groq_key
    )
```

## Generating Text

**Q2.** Write a function `generate_text(prompt)` that generates text with `max_tokens=10` and `temperature=0.7`. Test it with the prompt `"Once upon a time"`.

```python
def generate_text(prompt):
    response = client.chat.completions.create(
        
        temperature= 0.7,
        model= 'openai/gpt-oss-20b',
        messages = [
            {
                'role' : 'system',
                'content' : 'you will be given a prompt , complete it'
            },
           {
               'role': 'user',
               'content': 'Once upon a time'
           },
           {
               'role' : 'assistant',
               'content': 'Once upon a time, a little girl found a hidden door in the forest.'
           },
           {
                          'role': 'user',
                          'content': 'The sun rose over the mountains'
                      },
                      {
                          'role' : 'assistant',
                          'content': 'The sun rose over the mountains and painted the sky in soft gold.'
                      },
            {
                           'role': 'user',
                           'content': 'A new adventure begins'
                       },
                       {
                           'role' : 'assistant',
                           'content': 'A new adventure begins when you dare to take the first step.'
                       },
            {
                                       'role': 'user',
                                       'content': prompt
                                   },

        ]
    )
    
    return response.choices[0].message.content.strip()
```

```python
print(generate_text('funny world we live in '))
```

## Summarising Text

**Q4.** Write a function `text_summarizer(prompt)` that extracts a list of keywords from a block of text. Use a system message to define the task, and include a couple of example user/assistant exchanges showing text paired with its extracted keywords. Then test the function on a paragraph of your choice.

```python
def text_summarizer(prompt):
    response = client.chat.completions.create(
        model= 'openai/gpt-oss-20b',
        messages = [
            {
                'role' : 'system',
                'content' : 'you will be given a prompt , extract keywords from it'
            },
           {
               'role': 'user',
               'content': 'The quick brown fox jumps over the lazy dog.'
           },
           {
               'role' : 'assistant',
               'content': '["quick", "brown", "fox", "jumps", "lazy", "dog"]'
           },
           {
                          'role': 'user',
                          'content': "Artificial intelligence is transforming the healthcare industry."
                      },
                      {
                          'role' : 'assistant',
                          'content': '["artificial intelligence", "transforming", "healthcare", "industry"]'
                      },
            {
                           'role': 'user',
                           'content': "The city was crowded, noisy, and full of lights at night."
                       },
                       {
                           'role' : 'assistant',
                           'content': '["city", "crowded", "noisy", "lights", "night"]'
                       },
            {
                                       'role': 'user',
                                       'content': prompt
                                   },

        ]
    )
    return response.choices[0].message.content.strip()
```

```python
print(text_summarizer('slowly loosing my mind in a sea of thoughts'))
```

## Langchain

**Q6.** Using Langchain, build a simple question-answering system over a webpage of your choice:

- Load the page's content.
- Split the content into chunks.
- Create embeddings and store them in a vector store.
- Set up conversation memory.
- Build a conversational retrieval chain using a chat model.
- Ask the chain a question about the page's content and print the answer.

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationalRetrievalChain 
import re
```

```python
source_url = 'https://store.steampowered.com/app/50130/Mafia_II_Classic/'
```

```python
loader = WebBaseLoader(web_path= (source_url,))
raw_document = loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter()
splitted_documents = splitter.split_documents(raw_document)
```

```python
embeddings = HuggingFaceEmbeddings(
    model = 'sentence-transformers/all-MiniLM-L6-v2'
)
```

```python
vector_store = FAISS.from_documents(
    documents= splitted_documents,
    embedding= embeddings
)
```

```python
memory = ConversationBufferMemory(
    memory_key= 'chat_history',
    return_messages= True
)
```

```python
qa = ConversationalRetrievalChain.from_llm(
    ChatGroq(
        model = 'openai/gpt-oss-20b',
        api_key= groq_key,
        temperature= 0.8
    ),
    vector_store.as_retriever(),
    memory = memory
)
```

```python
query = 'whats the name of game in given documents ?'
```

```python
result = qa.invoke(
    {
        'question' : query
    }
)
```

```python
clean_text = re.sub(r'\s+', ' ', result['answer']).strip()
print(clean_text)
```
