```python
from openai import OpenAI
import config

client = OpenAI(
    api_key= config.api_key,
    base_url="https://api.groq.com/openai/v1/"
)

def text_generate(prompt):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=100,
        temperature=0,
    )
    return response.choices[0].message.content

print(text_generate("sad"))
```

```python
def txt_generate(prompt):
    response = client.chat.completions.create(
        model = 'openai/gpt-oss-20b',
        max_tokens= 100,
        messages= [{
            'role': 'user', 'content': prompt
        }],
        temperature= 2
    ) 
    return response.choices[0].message.content


print(txt_generate('very angry')) 
```

```python
def txt_summarizer(prompt):
    response = client.chat.completions.create(
        messages= [
            {
            'role' : 'system',
            'content' : 'you will be given a text , extract important words from it'
        },
        {
            'role' : 'user' ,
            'content' : 'Renewable energy is becoming increasingly important as countries seek cleaner ways to power homes, businesses, and transportation. Solar panels, wind turbines, and hydroelectric systems reduce reliance on fossil fuels and help limit carbon emissions.'
        },
        {
            'role' : 'assistant',
            'content': 'Important words: renewable energy, cleaner, solar panels, wind turbines, hydroelectric systems, fossil fuels, carbon emissions.'
        },
        {
            'role': 'user',
            'content' : prompt
        }
        ],
        model = 'openai/gpt-oss-20b',
        max_tokens= 100,
        temperature= 1
    )
    return response.choices[0].message.content
    
```

```python
print(txt_summarizer('Learning a new skill takes patience, practice, and curiosity. Small daily efforts can gradually build confidence and turn difficult tasks into familiar habits.'))
```

```python
def genre(prompt):
    response = client.chat.completions.create(
        messages= [
            {
            'role' : 'system',
            'content' : 'you will be given a game name , define its genre'
        },
        {
            'role' : 'user' ,
            'content' : 'beyond two souls'
        },
        {
            'role' : 'assistant',
            'content': 'Interactive drama, supernatural, adventure, thriller'
        },
        
        {
            'role' : 'user',
            'content': 'me hun na '
        },
        {
            'role' : 'assistant',
            'content': 'sorry I cant find a game with this name'
        },
        {
                    'role': 'user',
                    'content' : prompt
                },
        ],
        model = 'openai/gpt-oss-20b',
        max_tokens=256,
        temperature= 0.7
    )
    return response.choices[0].message.content.strip()
```

```python
print(genre('cs 2'))
```
