```python
import os
from pathlib import Path
from dotenv import load_dotenv

env_file = Path.cwd() / "secrets.env"
load_dotenv(env_file)

groq_key = os.getenv("GROQ_API_KEY")
openai_key = os.getenv("OPENAI_API_KEY")

if not groq_key:
    raise RuntimeError("GROQ_API_KEY is missing. Add it to secrets.env, then restart and run this cell.")
```

```python
from openai import OpenAI
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationalRetrievalChain
from langchain_groq import ChatGroq
```

```python
client = OpenAI(
    api_key=groq_key,
    base_url="https://api.groq.com/openai/v1/"
)

def text_generate(prompt):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=100,
        temperature=0,
    )
    return response.choices[0].message.content

print(text_generate("sad"))
```

```python
def txt_generate(prompt):
    response = client.chat.completions.create(
        model = 'openai/gpt-oss-20b',
        max_tokens= 100,
        messages= [{
            'role': 'user', 'content': prompt
        }],
        temperature= 2
    ) 
    return response.choices[0].message.content


print(txt_generate('very angry')) 
```

```python
def txt_summarizer(prompt):
    response = client.chat.completions.create(
        messages= [
            {
            'role' : 'system',
            'content' : 'you will be given a text , extract important words from it'
        },
        {
            'role' : 'user' ,
            'content' : 'Renewable energy is becoming increasingly important as countries seek cleaner ways to power homes, businesses, and transportation. Solar panels, wind turbines, and hydroelectric systems reduce reliance on fossil fuels and help limit carbon emissions.'
        },
        {
            'role' : 'assistant',
            'content': 'Important words: renewable energy, cleaner, solar panels, wind turbines, hydroelectric systems, fossil fuels, carbon emissions.'
        },
        {
            'role': 'user',
            'content' : prompt
        }
        ],
        model = 'openai/gpt-oss-20b',
        max_tokens= 100,
        temperature= 1
    )
    return response.choices[0].message.content
    
```

```python
print(txt_summarizer('Learning a new skill takes patience, practice, and curiosity. Small daily efforts can gradually build confidence and turn difficult tasks into familiar habits.'))
```

```python
def genre(prompt):
    response = client.chat.completions.create(
        messages= [
            {
            'role' : 'system',
            'content' : 'you will be given a game name , define its genre'
        },
        {
            'role' : 'user' ,
            'content' : 'beyond two souls'
        },
        {
            'role' : 'assistant',
            'content': 'Interactive drama, supernatural, adventure, thriller'
        },
        
        {
            'role' : 'user',
            'content': 'me hun na '
        },
        {
            'role' : 'assistant',
            'content': 'sorry I cant find a game with this name'
        },
        {
                    'role': 'user',
                    'content' : prompt
                },
        ],
        model = 'openai/gpt-oss-20b',
        max_tokens=256,
        temperature= 0.7
    )
    return response.choices[0].message.content.strip()
```

```python
print(genre('cs 2'))
```

```python
source_url = 'https://www.gutenberg.org/cache/epub/1342/pg1342.txt'
loader = WebBaseLoader(web_paths=(source_url,))
```

```python
raw_documents = loader.load()
page_text = raw_documents[0].page_content
print(raw_documents[0].metadata.get('source'))
print(f'Loaded {len(page_text):,} characters.')
print(page_text[:1_000])

if 'author: jane austen' not in page_text.lower():
    raise RuntimeError('The expected document was not loaded; do not run the retrieval step yet.')
```

```python
text_splitter = RecursiveCharacterTextSplitter()
split_documents = text_splitter.split_documents(raw_documents)
```

```python
embeddings = HuggingFaceEmbeddings(
     model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorStore = FAISS.from_documents(documents=split_documents, embedding= embeddings)
```

```python
memory = ConversationBufferMemory(memory_key= 'chat_history', return_messages= True)
```

```python
qa = ConversationalRetrievalChain.from_llm(
    ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=groq_key,
    temperature=0,
),
vectorStore.as_retriever(),
memory = memory
)
```

```python
query = 'Who is the author of Pride and Prejudice?'
```

```python
result = qa.invoke({
    'question': query
})

for i, document in enumerate(vectorStore.similarity_search(query, k=3), start=1):
    print(f'\n--- Retrieved chunk {i} ---')
    print(document.page_content[:800])
```
