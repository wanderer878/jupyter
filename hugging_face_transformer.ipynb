```python
model = 'bert-base-uncased'
sentence = 'Mafia 2 is a really good game, but I prefer Mafia 3.'
```

```python
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
input_ids = tokenizer(sentence)
print(input_ids)
```

```python
tokens = tokenizer.tokenize(sentence)
print(tokens)
```

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
```

```python
tokens_back = tokenizer.decode(token_ids)
print(tokens_back)
```

```python
print(tokenizer.decode(101))
print(tokenizer.decode(102))
```

```python
model = "distilbert-base-uncased-finetuned-sst-2-english"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
token_type = tokenizer(sentence, return_tensors='pt')
print(token_type)
```

```python
model = AutoModelForSequenceClassification.from_pretrained('distilbert-base-uncased-finetuned-sst-2-english')
```

```python
with torch.no_grad():
    logits = model(**token_type).logits

score = logits.argmax().item()
print(score)
model.config.id2label[score]
```

```python
model_directory = 'my_saved_model'

tokenizer.save_pretrained(model_directory)
model.save_pretrained(model_directory)
```
